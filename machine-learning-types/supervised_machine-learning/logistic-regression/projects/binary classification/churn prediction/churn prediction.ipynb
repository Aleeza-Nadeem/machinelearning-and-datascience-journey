#` The Core Problem`
High-value customers are churning early in their relationship with the company due to an expectation-experience gap, zero contract lock-in, and friction in product setup.

The Expectation-Support Gap: Customers buying high-speed Fiber Optic internet pay premium rates ($91.50/month average), but over 70% of them lack essential support services like TechSupport or OnlineSecurity. When minor setup issues or speed drops occur, they feel cheated by the high price tag.

Low Switching Barriers: A majority of new customers start on flexible Month-to-Month contracts paired with manual Electronic Check payments. This combination gives frustrated users an easy, low-friction opportunity to leave at any moment.

Onboarding Drop-off: New customers are leaving before realizing the full value of the service, with over 55% of total churn occurring within the first 12 months.

#`The Effect of the Problem`
The impact of this high churn rate extends far beyond simple customer numbers:

Massive Revenue Bleed: Out of 7,043 total customers, 1,869 (26.54%) have left. This represents a monthly revenue loss of $139,130.85—meaning the company is losing 30.5% of its total monthly recurring revenue.

Disproportionate High-Value Loss: Premium Fiber Optic customers account for 82.15% ($114,300) of all lost revenue. The company is retaining its low-margin users (DSL/Landline) while losing its most profitable segment.

Wasted Acquisition Costs (CAC): Acquiring premium internet subscribers requires heavy upfront sales and installation expenses. Losing 47.4% of new customers in Year 1 means the company fails to recoup its customer acquisition costs.

#`outlier checking in categoricaal col `

```python
# Select categorical columns (excluding customerID)
categorical_cols = df.select_dtypes(include=['object']).columns.drop('customerID')

def check_categorical_outliers(df, cols, rare_threshold=0.01):
    print("=" * 60)
    print("CATEGORICAL OUTLIER & UNIQUE VALUE ANALYSIS")
    print("=" * 60)

    for col in cols:
        counts = df[col].value_counts(dropna=False)
        proportions = df[col].value_counts(normalize=True, dropna=False) * 100

        # Combine count and percentage into a summary table
        summary = pd.DataFrame({'Count': counts, 'Percentage (%)': proportions.round(2)})

        print(f"\nFeature: '{col}' | Unique Values Count: {df[col].nunique()}")
        print("-" * 45)
        print(summary)

        # Flag rare categories (< 1% representation)
        rare_cats = proportions[proportions < (rare_threshold * 100)]
        if not rare_cats.empty:
            print(f"⚠️ RARE CATEGORIES DETECTED (< {rare_threshold*100}%): {list(rare_cats.index)}")

# --- Run Categorical Outlier Check ---
check_categorical_outliers(df, categorical_cols, rare_threshold=0.01)
```

#`outlier checking and capping using percentile method `

```python

# 1. Load Data and Clean TotalCharges
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn (1).csv')
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].str.strip(), errors='coerce').fillna(0)

# Select numeric columns
numeric_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']

# 2. Function to Detect Outliers using Percentiles (1st & 99th Percentile)
def detect_percentile_outliers(df, cols, lower_pct=0.01, upper_pct=0.99):
    print("=" * 60)
    print(f"PERCENTILE OUTLIER DETECTION ({lower_pct*100}th to {upper_pct*100}th Percentile)")
    print("=" * 60)

    thresholds = {}
    for col in cols:
        lower_bound = df[col].quantile(lower_pct)
        upper_bound = df[col].quantile(upper_pct)
        thresholds[col] = (lower_bound, upper_bound)

        below_count = (df[col] < lower_bound).sum()
        above_count = (df[col] > upper_bound).sum()

        print(f"Column: '{col}'")
        print(f"  - Lower Bound ({lower_pct*100:.0f}%): {lower_bound:.2f} | Count Below: {below_count}")
        print(f"  - Upper Bound ({upper_pct*100:.0f}%): {upper_bound:.2f} | Count Above: {above_count}\n")

    return thresholds

# # 3. Function to Cap (Clip) Numerical Outliers
# def cap_outliers_percentile(df, cols, thresholds):
#     df_capped = df.copy()
#     for col in cols:
#         lower_bound, upper_bound = thresholds[col]
#         # Clip values below lower_bound to lower_bound, and above upper_bound to upper_bound
#         df_capped[col] = np.clip(df_capped[col], lower_bound, upper_bound)
#     return df_capped

# # --- Run Percentile Detection & Capping ---
# thresholds = detect_percentile_outliers(df, numeric_cols, lower_pct=0.01, upper_pct=0.99)
# df_capped = cap_outliers_percentile(df, numeric_cols, thresholds)

# # Verify max value before and after capping
# print("--- Capping Verification ---")
# print(f"MonthlyCharges Max Before Capping: ${df['MonthlyCharges'].max():.2f}")
# print(f"MonthlyCharges Max After Capping:  ${df_capped['MonthlyCharges'].max():.2f}")
```

```python
# Function to Cap (Clip) Numerical Outliers
def cap_outliers_percentile(df, cols, thresholds):
    df_capped = df.copy()
    for col in cols:
        lower_bound, upper_bound = thresholds[col]
        # Clip values below lower_bound to lower_bound, and above upper_bound to upper_bound
        df_capped[col] = np.clip(df_capped[col], lower_bound, upper_bound)
    return df_capped
```

```python
# --- Run Percentile Detection & Capping ---
thresholds = detect_percentile_outliers(df, numeric_cols, lower_pct=0.01, upper_pct=0.99)
df_capped = cap_outliers_percentile(df, numeric_cols, thresholds)

# Verify max value before and after capping
print("--- Capping Verification ---")
print(f"MonthlyCharges Max Before Capping: ${df['MonthlyCharges'].max():.2f}")
print(f"MonthlyCharges Max After Capping:  ${df_capped['MonthlyCharges'].max():.2f}")
```

#`Onboarding Danger Zone" (Tenure Thresholds)`
If you plot tenure against Churn, you don't just see a smooth line—you see two distinct customer populations:

Months 0–12: Peak churn intensity (over 40% of new sign-ups leave within their first year).

Months 24+: Churn drops drastically.

Months 60–72: Churn is practically zero (< 10%).

`Deep Dive Insight`:
If a customer survives the first 12 months, their lifetime value (LTV) increases exponentially. Retention strategies and promotional discounts should be heavily focused on the first 0–6 months to get customers past the critical drop-off zone.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data and prepare churn variable
df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn (1).csv')
df['Churn_Num'] = (df['Churn'] == 'Yes').astype(int)

# =========================================================
# 1. CATEGORIZE PAYMENT METHODS (Manual vs Automatic)
# =========================================================
# Map each payment method into either 'Manual' or 'Automatic'
payment_type_map = {
    'Electronic check': 'Manual (Electronic Check)',
    'Mailed check': 'Manual (Mailed Check)',
    'Bank transfer (automatic)': 'Automatic (Bank Transfer)',
    'Credit card (automatic)': 'Automatic (Credit Card)'
}

df['Payment_Category'] = df['PaymentMethod'].map(payment_type_map)
df['Is_Automatic'] = df['PaymentMethod'].str.contains('automatic').map({True: 'Automatic', False: 'Manual'})

# =========================================================
# 2. CALCULATE SUMMARY STATS BY PAYMENT METHOD
# =========================================================
print("=== CHURN RATE BY EXACT PAYMENT METHOD ===")
payment_stats = df.groupby('PaymentMethod')['Churn_Num'].agg(
    Total_Customers='count',
    Churned_Customers='sum',
    Churn_Rate_Pct=lambda x: round(x.mean() * 100, 2)
).reset_index().sort_values(by='Churn_Rate_Pct', ascending=False)

print(payment_stats.to_string(index=False))

print("\n=== CHURN RATE: MANUAL VS AUTOMATIC PAYMENTS ===")
auto_vs_manual = df.groupby('Is_Automatic')['Churn_Num'].agg(
    Total_Customers='count',
    Churned_Customers='sum',
    Churn_Rate_Pct=lambda x: round(x.mean() * 100, 2)
).reset_index()

print(auto_vs_manual.to_string(index=False))

# =========================================================
# 3. VISUALIZE THE PAYMENT DISCONNECT
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Plot 1: Breakdown by all 4 payment methods
sns.barplot(
    data=payment_stats,
    x='Churn_Rate_Pct',
    y='PaymentMethod',
    palette=['#e74c3c' if 'Electronic' in x else '#3498db' for x in payment_stats['PaymentMethod']],
    ax=axes[0]
)
axes[0].set_title('Churn Rate by Specific Payment Method', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Churn Rate (%)')
axes[0].set_ylabel('')

# Add percent labels on top of bars
for p in axes[0].patches:
    width = p.get_width()
    axes[0].annotate(f'{width:.1f}%',
                     (width + 1, p.get_y() + p.get_height() / 2.),
                     va='center', fontsize=10, fontweight='bold')

# Plot 2: High-Level Comparison (Manual vs Automatic)
sns.barplot(
    data=auto_vs_manual,
    x='Is_Automatic',
    y='Churn_Rate_Pct',
    palette=['#e74c3c', '#2ecc71'],
    ax=axes[1]
)
axes[1].set_title('Overall: Manual vs Automatic Payment Churn Rate', fontsize=12, fontweight='bold')
axes[1].set_ylabel('Churn Rate (%)')
axes[1].set_xlabel('')
axes[1].set_ylim(0, 50)

# Add percent labels
for p in axes[1].patches:
    height = p.get_height()
    axes[1].annotate(f'{height:.1f}%',
                     (p.get_x() + p.get_width() / 2., height + 1),
                     ha='center', va='center', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

# =========================================================
# 1. BASE CLEANING & LOG TRANSFORMATION (df_processed)
# =========================================================


# Extract target (y) and baseline features (X)
y = (df_processed['Churn'] == 'Yes').astype(int)
X = df_processed.drop(columns=['customerID', 'Churn', 'TotalCharges'])

# =========================================================
# 2. STRING CLEANING & ONE-HOT ENCODING (Safe Pre-Split)
# =========================================================
# Cleaning string values and OHE on discrete categories doesn't leak numerical distributions
cat_cols = X.select_dtypes(include=['object']).columns

for col in cat_cols:
    X[col] = X[col].astype(str).str.replace('-', '_').str.replace(' ', '_')

X_encoded = pd.get_dummies(X, columns=cat_cols, drop_first=True, dtype=int)

# =========================================================
# 3. TRAIN-TEST SPLIT (MUST HAPPEN BEFORE SCALING!)
# =========================================================
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

print(f"✓ Split Complete — Train: {X_train.shape[0]} samples | Test: {X_test.shape[0]} samples")

# =========================================================
# 4. PREVENT LEAKAGE: FIT SCALER ON TRAIN ONLY
# =========================================================
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges_log']

scaler = StandardScaler()

# Fit parameters (mean, std) ONLY on X_train
X_train[num_cols] = scaler.fit_transform(X_train[num_cols])

# Transform X_test using X_train's fitted parameters
X_test[num_cols] = scaler.transform(X_test[num_cols])

print("✓ Standardization complete without Data Leakage!")

# =========================================================
# 5. TRAIN LOGISTIC REGRESSION MODEL
# =========================================================
model = LogisticRegression(class_weight='balanced',max_iter=1000, random_state=42)
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print("\n=== FINAL PROFESSIONAL PIPELINE RESULTS ===")
print(classification_report(y_test, y_pred))
print(f"ROC-AUC Score: {roc_auc_score(y_test, y_proba):.4f}")
```

#`Model Transition Rationale (Logistic Regression $\rightarrow$ Random Forest)`

🚨` Why We Maxed Out Logistic RegressionDuring our experimentation with Logistic Regression`, we applied several optimization techniques to resolve the performance gap on minority churn predictions (Class 1):

- `Adjusting Class Weights (class_weight='balanced')`:Result: Increased Class 1 Recall ($0.53 \rightarrow 0.80$), but dropped Precision down to $0.51$, causing a high volume of false alarms.

- `Threshold Tuning ($0.30 - 0.50$)`:Result: Highlighted a strict 1-to-1 Precision-Recall Trade-off. Raising the threshold reduced false alarms (higher Precision) but caused genuine churners to be missed (lower Recall).

- ` Hyperparameter Grid Search` (GridSearchCV):Result: Optimizing $C$, solvers, and penalty terms resulted in marginal improvements, plateauing around ~0.56 Precision and ~0.73 Recall

- ` The Mathematical LimitationLogistic Regression` is a linear model. It attempts to separate churners from non-churners using a single linear hyper-plane.However, customer churn behavior in this dataset is non-linear and driven by complex interactions between multiple variables (e.g., a customer with Fiber Optic AND no Tech Support AND a Month-to-Month contract exhibits exponential churn risk compared to someone with only one of those features). A linear boundary simply cannot capture these non-linear decision spaces without sacrificing either Precision or Recall.🚀 Strategic Shift to Ensemble Learning (Random Forest)To lift both Precision and Recall simultaneously without settling for a trade-off, we are moving to Random Forest Classification.
